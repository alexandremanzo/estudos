```python
# pip install google-cloud-bigquery
# pip install pandas_gbq

from google.cloud import bigquery
import pandas as pd
```

```python
import os

# caminho da chave da conta de servico, fora do repositorio
credencial = os.environ["GOOGLE_APPLICATION_CREDENTIALS"]
```

```python
client = bigquery.Client.from_service_account_json(credencial)
```

```python
query = """
SELECT 
data,
`Tipo Movimentação` as Tipo,
`Valor trat` as Valor

 FROM `seu-projeto.seu_dataset.sua_tabela` 

"""
```

```python
resultado = client.query(query)
```

```python
#resultado formatado
import pandas as pd

# Suponha que resultado seja um iterável de Row
dados = [dict(row) for row in resultado]
df = pd.DataFrame(dados)
df.head()
#print(df)
```
