```python
from transformers import pipeline
import pandas as pd

# Classificador de sentimentos
sentiment = pipeline("sentiment-analysis")

# Dados de exemplo
textos = [
    "Eu amei esse produto!",
    "Horrível, nunca mais compro.",
    "Achei ok, podia ser melhor.",
    "Fantástico, parabéns!",
    "Não gostei."
]

# Classificar
resultados = sentiment(textos)

# Exibir resultados em DataFrame
df = pd.DataFrame({
    "Texto": textos,
    "Sentimento": [r['label'] for r in resultados],
    "Confiança": [round(r['score'], 3) for r in resultados]
})

print(df)
```
