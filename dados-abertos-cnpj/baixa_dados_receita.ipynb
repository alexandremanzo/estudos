```python
import os
import requests
from datetime import datetime
import zipfile

# === CONFIGURAÇÕES ===
pasta_download = "C:/DownloadsCNPJ"
pasta_extracao_base = "C:/DadosExtraidos"

# === DATA ATUAL ===
data = datetime.now()
ano = data.year
mes = f"{data.month:02d}"

# === NOME DE ARQUIVO E URL ===
nome_arquivo_zip = f"Cnaes-{ano}-{mes}.zip"
url = f"https://arquivos.receitafederal.gov.br/dados/cnpj/dados_abertos_cnpj/{ano}-{mes}/Cnaes.zip"

# === CAMINHOS COMPLETOS ===
os.makedirs(pasta_download, exist_ok=True)
caminho_zip = os.path.join(pasta_download, nome_arquivo_zip)

pasta_extracao = os.path.join(pasta_extracao_base, f"{ano}-{mes}")
os.makedirs(pasta_extracao, exist_ok=True)

# === FAZ O DOWNLOAD ===
print("🔽 Baixando o arquivo...")
resposta = requests.get(url)
with open(caminho_zip, "wb") as f:
    f.write(resposta.content)
print(f"✅ Arquivo salvo: {caminho_zip}")

# === EXTRAI O CONTEÚDO ===
print("📦 Extraindo o conteúdo...")
with zipfile.ZipFile(caminho_zip, 'r') as zip_ref:
    zip_ref.extractall(pasta_extracao)
print(f"✅ Arquivos extraídos para: {pasta_extracao}")

# === REMOVE O .ZIP ===
os.remove(caminho_zip)
print("🧹 Arquivo .zip removido após extração.")


```

```python
## Faz:
## Baixa todos esses arquivos com base no ano e mês atual
## Salva com nomes no padrão: Empresas-2025-05.zip
## Extrai cada ZIP para: C:/DadosReceita/2025-05/Empresas/, .../Cnaes/ etc.

import os
import requests
from datetime import datetime
import zipfile

# === CONFIGURAÇÕES ===
PASTA_BASE = "C:/DadosReceita"  # Altere se quiser
ARQUIVOS = [
    "Empresas",
    "Estabelecimentos",
    "Socios",
    "Cnaes",
    "Municipios",
    "Naturezas",
    "Qualificacoes",
    "Paises",
    "Motivos"
]

# === DATA ATUAL ===
data = datetime.now()
ano = data.year
mes = f"{data.month:02d}"
pasta_mes = f"{ano}-{mes}"

# === CRIA PASTA BASE DO MÊS ===
caminho_base_mes = os.path.join(PASTA_BASE, pasta_mes)
os.makedirs(caminho_base_mes, exist_ok=True)

# === LOOP PARA CADA ARQUIVO ===
for nome in ARQUIVOS:
    nome_zip = f"{nome}-{pasta_mes}.zip"
    url = f"https://arquivos.receitafederal.gov.br/dados/cnpj/dados_abertos_cnpj/{pasta_mes}/{nome}.zip"
    caminho_zip = os.path.join(caminho_base_mes, nome_zip)
    pasta_extraida = os.path.join(caminho_base_mes, nome)

    try:
        print(f"🔽 Baixando {nome}...")
        resposta = requests.get(url)
        if resposta.status_code == 200:
            with open(caminho_zip, "wb") as f:
                f.write(resposta.content)
            print(f"✅ Download de {nome_zip} concluído.")

            # Extrair
            print(f"📦 Extraindo {nome_zip}...")
            os.makedirs(pasta_extraida, exist_ok=True)
            with zipfile.ZipFile(caminho_zip, 'r') as zip_ref:
                zip_ref.extractall(pasta_extraida)
            print(f"✅ Arquivo extraído para {pasta_extraida}")

            # Excluir ZIP
            os.remove(caminho_zip)
            print(f"🧹 ZIP {nome_zip} removido.\n")

        else:
            print(f"⚠️ Falha ao baixar {nome}. Código HTTP: {resposta.status_code}\n")

    except Exception as e:
        print(f"❌ Erro ao processar {nome}: {e}\n")
```
