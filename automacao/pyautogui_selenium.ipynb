```python
# Apertar a tecla do windows e seleciona algo
import pyautogui

pyautogui.PAUSE = 2
pyautogui.press ("Win")
pyautogui.write ("google chrome")
pyautogui.press("backaspace")

pyautogui.PAUSE = 20
pyautogui.press("enter")
pyautogui.click(640,685)

pyautogui.PAUSE = 5
pyautogui.write ("www.youtube.com")
pyautogui.press("backaspace")
pyautogui.press("enter")
pyautogui.click(938,1204)
pyautogui.press("enter")
pyautogui.write ("São Paulo Futebol Clube")
pyautogui.press("enter")
```
